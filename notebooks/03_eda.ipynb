```python
import plotly.express as px
```

```python
import pandas as pd
df = pd.read_csv("../data/dataset.csv")
```

```python
#delivery status distribution
status_counts = (
    df["delivery_status"]
    .fillna("unknown")
    .value_counts()
    .rename_axis("delivery_status")
    .reset_index(name="trip_count")
)

fig = px.bar(
    status_counts,
    x="delivery_status",
    y="trip_count",
    title="Distribution of Delivery Status"
)
fig.show()
```

```python
#distance vs fuel consumption
fig = px.scatter(
    df,
    x="distance_km",
    y="fuel_L",
    color="delivery_status",
    hover_data=["vehicle_id", "trip_id"],
    title="Distance vs Fuel Consumption"
)
fig.show()
```

```python
#fuel efficiency by vehicle
valid = df[
    df["distance_km"].gt(0) &
    df["fuel_L"].ge(0)
].copy()

valid["fuel_L_per_100km"] = (
    valid["fuel_L"] / valid["distance_km"] * 100
)

vehicle_fuel = (
    valid.groupby("vehicle_id", dropna=False)
    .agg(
        trips=("trip_id", "count"),
        distance_km=("distance_km", "sum"),
        fuel_L=("fuel_L", "sum")
    )
    .reset_index()
)

vehicle_fuel["fuel_L_per_100km"] = (
    vehicle_fuel["fuel_L"] /
    vehicle_fuel["distance_km"] * 100
)

fig = px.bar(
    vehicle_fuel,
    x="vehicle_id",
    y="fuel_L_per_100km",
    title="Fuel Consumption per 100 km by Vehicle"
)
fig.show()
```

```python
#trip distance distribution
fig = px.histogram(
    df,
    x="distance_km",
    nbins=30,
    title="Distribution of Trip Distances"
)
fig.show()
```

```python
#score distribution
fig = px.histogram(
    df,
    x="driver_score",
    nbins=20,
    title="Distribution of driver Scores"
)
fig.show()
```
