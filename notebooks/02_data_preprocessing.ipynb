```python
import pandas as pd
```

```python
df = pd.read_csv("../data/dataset.csv")
```

```python
print(df.isnull().sum())
```

```python
#data standarization
df.columns = df.columns.str.strip()

for col in ["trip_id", "vehicle_id", "driver_id"]:
    df[col] = df[col].astype("string").str.strip()

for col in ["origin", "destination", "delivery_status"]:
    df[col] = df[col].astype("string").str.strip()

for col in ["distance_km", "fuel_L", "driver_score"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df["delivery_status"] = df["delivery_status"].str.lower()
```

```python
#checking invalid numerical values
print("Non-positive distances:",
      (df["distance_km"] <= 0).sum())

print("Negative fuel readings:",
      (df["fuel_L"] < 0).sum())

print("Missing trip IDs:", df["trip_id"].isna().sum())
```

```python
#removing duplicate intervals
df = df.drop_duplicates()

# Only if trip_id is supposed to identify one unique trip:
# Investigate repeated IDs before resolving them.
print("Repeated trip IDs:",
      df["trip_id"].duplicated().sum())
```

```python
#saving cleaned sataset
df.to_csv("../data/dataset.csv", index=False)

print("Cleaned dataset saved.")
print("Final shape:", df.shape)
```
